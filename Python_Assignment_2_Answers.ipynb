# 1. List Creation

## Create age_list

```python
age_list = [24, 25, 26, 27, 28]

print(age_list)
```

## Create name_list

```python
name_list = ["Akash", "Anu", "Arun", "Meera", "Rahul"]

print(name_list)
```

# 2. List Operations / Modifications

## Append "Yazhini"

```python
name_list.append("Yazhini")

print(name_list)
```

## Insert 30 at Index 2

```python
age_list.insert(2, 30)

print(age_list)
```

## Remove "Yazhini"

```python
name_list.remove("Yazhini")

print(name_list)
```

## Pop the Last Element

```python
age_list.pop()

print(age_list)
```

## Extend age_list

```python
age_list.extend([29, 30, 26])

print(age_list)
```

## Sort in Descending Order

```python
age_list.sort(reverse=True)

print(age_list)
```

## Maximum, Minimum and Sum

```python
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of all ages:", sum(age_list))
```

## Print Values

```python
print(student_marks.items())
```

# 5. Sets

## Create my_set

```python
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}

print(my_set)
```

## Explanation

A set contains only unique values, so duplicate vowels are removed. Sets are unordered, so the displayed order may vary.

## Attempt to Modify a Set Element

```python
my_set = {'a', 'e', 'i', 'o', 'u'}

my_set[4] = 's'
```

## Explanation

This raises a TypeError because sets do not support indexing or item assignment. Use `my_set.add('s')` to add an element.

## Create set1 and set2

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)
```

## Union

```python
print(set1.union(set2))
```

## Intersection

```python
print(set1.intersection(set2))
```

# 6. Operators & Conditional Statements

## Performance Category Program

```python
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")
elif score > 7:
    print("Above Average: Excellent performance! Keep it up.")
elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")
else:
    print("Below Average: Need to improve your performance, consistent practice will lead to better results.")
```
