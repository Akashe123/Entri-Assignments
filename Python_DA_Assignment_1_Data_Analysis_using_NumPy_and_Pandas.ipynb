## 1. NumPy Array Operations

### 1.1 Import NumPy

```python
import numpy as np
```

### 1.2 Create a 1D NumPy Array

Create `temperatures_w1` for Week 1 using the temperatures provided in the question.

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

### 1.3 Inspect Shape, Data Type, and Number of Elements

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

### 1.4 Convert Celsius to Fahrenheit

Formula: **Fahrenheit = (Celsius × 9/5) + 32**

```python
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:")
print(temperatures_f)
```

### 1.5 Find Maximum, Minimum, and Mean Temperature

```python
maximum_temperature = np.max(temperatures_w1)
minimum_temperature = np.min(temperatures_w1)
mean_temperature = np.mean(temperatures_w1)

print("Maximum temperature:", maximum_temperature)
print("Minimum temperature:", minimum_temperature)
print("Mean temperature:", round(mean_temperature, 2))
```

### 1.6 Array Slicing and Indexing

Extract the first three days, weekend temperatures, and middle three days.

```python
first_three_days = temperatures_w1[:3]
weekend = temperatures_w1[-2:]
middle_three_days = temperatures_w1[2:5]

print("First three days:", first_three_days)
print("Weekend:", weekend)
print("Middle three days:", middle_three_days)
```

## 2. Create and Analyze a 2D NumPy Array

Each row represents the temperatures for one week.

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

temperatures
```

### 2.1 Inspect Shape, Data Type, and Total Number of Elements

```python
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)
```

### 2.2 Extract Each Week and the Weekend Temperatures

```python
week1 = temperatures[0]
week2 = temperatures[1]
week1_weekend = temperatures[0, -2:]
week2_weekend = temperatures[1, -2:]

print("Week 1:", week1)
print("Week 2:", week2)
print("Week 1 weekend:", week1_weekend)
print("Week 2 weekend:", week2_weekend)
```

# 3. Pandas Series

## 3.1 Import Pandas and Create the Series

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

marks
```

## 3.2 Indexing and Slicing

```python
print("Mark of 1st rank student:", marks.iloc[0])

print("Top 3 ranks using loc:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

print("Mark of 3rd rank student using iloc:", marks.iloc[2])

print("Ranks with marks greater than 90:")
print(marks[marks > 90])
```

## 3.3 Manipulating the Series

Modify Rank1, remove Rank5, and calculate CGPA by dividing each mark by 10.

```python
marks.loc['Rank1'] = 100
marks = marks.drop('Rank5')
cgpa = marks / 10

print("Updated marks:")
print(marks)

print("CGPA:")
print(cgpa)
```

# 4. Pandas DataFrame

## 4.1 Create the Transactions DataFrame

```python
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

transactions
```

## 4.2 Data Exploration

```python
transactions
```

### Display the Head and Tail

```python
print("First 5 rows:")
display(transactions.head())

print("Last 5 rows:")
display(transactions.tail())
```

```python
print("Shape:", transactions.shape)
print("Column names:", list(transactions.columns))
print("Data types:")
print(transactions.dtypes)
```

## 4.3 Manipulating the DataFrame

### Modify TransactionID 102

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
transactions
```

### Add a Discount Column

The discount is calculated as 10% of the Amount.

```python
transactions['Discount'] = transactions['Amount'] * 0.10
transactions
```
