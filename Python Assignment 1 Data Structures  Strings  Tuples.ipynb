# 3.String Methods

## Create the String

```python
strM = "Python beginner tutorial"

print(strM)
```

## Uppercase

```python
print(strM.upper())
```

## Lowercase

```python
print(strM.lower())
```

## Capitalize

```python
print(strM.capitalize())
```

## Count 't'

```python
print(strM.count('t'))
```

## Replace Python

```python
print(strM.replace("Python", "Machine Learning"))
```

# 4. Tuples

## Create Tuples

```python
t1 = (10, 20, 30)
t2 = (40, 50, 60)

print(t1)
print(t2)
```

## Concatenate Tuples

```python
t_combine = t1 + t2

print(t_combine)
```

## Repeat 3 Times

```python
print(t_combine * 3)
```

## Access 3rd Element

```python
print(t_combine[2])
```

## First Three Elements

```python
print(t_combine[:3])
```

## Last Three Elements

```python
print(t_combine[-3:])
```
